```python
chars = set()
label_path = "/kaggle/input/datasets/bindang/lao-plate/laos_plate_dataset/train.txt"
with open(label_path, encoding="utf-8") as f:
    for line in f:
        parts = line.strip().split(maxsplit=1)
        if len(parts) < 2:
            continue
        text = parts[1]
        for c in text:
            chars.add(c)
chars = sorted(chars)
save_path = "/kaggle/working/PaddleOCR/laos_dict.txt"
with open(save_path, "w", encoding="utf-8") as f:
    for c in chars:
        f.write(c + "\n")
print("saved:", save_path)
print("num chars:", len(chars))
print(chars)
```
